```python
from PyPDF2 import PdfReader, PdfWriter
from pdf2image import convert_from_path
import os
import time

```

截取pdf生成新文件

```python
def extract_pages(pdf_path, output_pdf_path, max_pages):
    reader = PdfReader(pdf_path)
    writer = PdfWriter()

    for i in range(min(max_pages, len(reader.pages))):
        writer.add_page(reader.pages[i])

    with open(output_pdf_path, 'wb') as output_pdf:
        writer.write(output_pdf)
```

将pdf转换成图片

```python
def pdf_to_images(pdf_path, output_folder, dpi=300):
    start_time = time.time()

    # 确保输出文件夹存在
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # 将PDF转换成图片
    images = convert_from_path(pdf_path, dpi=dpi)

    for i, image in enumerate(images):
        # 保存每一页图片到输出文件夹
        image.save(os.path.join(output_folder, f'page_{i+1}.png'), 'PNG')

    end_time = time.time()
    print(f"处理时间: {end_time - start_time:.2f} 秒")
```

输入文件地址

```python
original_pdf_path = r'D:\Intelligent_Image_Recognition\pdf_img\origin_pdf\22-BB-096 02子项 04段冯楼北 桥梁工程(1).pdf'
extracted_pdf_path = r'D:\Intelligent_Image_Recognition\pdf_img\origin_pdf\extracted_2_5_pages.pdf'
output_folder = r'D:\Intelligent_Image_Recognition\pdf_img\results\res3'
```

执行

```python
# 提取前五页并生成新的 PDF 文件
extract_pages(original_pdf_path, extracted_pdf_path, max_pages=5)

# 将新的 PDF 文件转换成图片
pdf_to_images(extracted_pdf_path, output_folder)
```
